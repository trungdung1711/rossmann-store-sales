### 💩 Simple data cleaning, one file prototype

Customers
Number of customers that day
⚠️ Important:
Highly correlated with sales
BUT NOT available in test set
So you cannot use it as a feature in real prediction

👉 Think of it as “training-only insight”

Assortment
Product variety:
a = basic
b = extra
c = extended (largest variety)

Meaning:

More assortment → usually higher sales

CompetitionDistance
Distance to nearest competitor store (meters)

Meaning:

Closer competitor → usually lower sales

CompetitionOpenSinceMonth/Year
When competitor opened nearby

Meaning:

Helps model “competition pressure over time”

---

❌ Pitfall 1: Using closed stores

If Open = 0:

Sales = 0 always
But this is NOT real demand

👉 If you train on this:

Model learns “closed = zero sales”
This corrupts learning

❌ Pitfall 2: Using Customers in features
Customers strongly correlates with Sales
BUT NOT in test set

👉 If used:

your model becomes invalid for prediction

❌ Pitfall 3: Missing values are NOT random

For example:

Competition data missing ≠ error

It often means:

“no competitor nearby / unknown timing”

So naive filling can distort signal.

❌ Pitfall 4: Date leakage in feature engineering later

Even cleaning must prepare for:

time-based splits
correct sorting

```python
train_merged = train_merged[train_merged["Open"] == 1]
```

Why this is safe:

Because:

test set only includes open predictions implicitly
closed days are not meaningful demand signals

```python
train_merged = train_merged[train_merged["Sales"] > 0]
```

Why?
RMSPE ignores 0 sales anyway
zero sales often = closed or errors

#### Fix data type

Why?

Without this:

model treats categories as numbers
introduces fake ordering (0 < a < b < c)

```python
train_merged["StateHoliday"] = train_merged["StateHoliday"].astype(str)
test_merged["StateHoliday"] = test_merged["StateHoliday"].astype(str)
```

#### Competition distance

```python
median_dist = train_merged["CompetitionDistance"].median()

train_merged["CompetitionDistance"] = train_merged["CompetitionDistance"].fillna(median_dist)
test_merged["CompetitionDistance"] = test_merged["CompetitionDistance"].fillna(median_dist)
```

#### COmptetion start date

```python
for col in ["CompetitionOpenSinceMonth", "CompetitionOpenSinceYear"]:
    train_merged[col] = train_merged[col].fillna(0)
    test_merged[col] = test_merged[col].fillna(0)
```

#### Promt2 field

```python
for col in ["Promo2SinceWeek", "Promo2SinceYear", "PromoInterval"]:
    train_merged[col] = train_merged[col].fillna(0)
    test_merged[col] = test_merged[col].fillna(0)
```

Raw Data
   ↓
(1) Light EDA  ← initial understanding (VERY IMPORTANT)
   ↓
(2) Light cleaning (fix obvious issues)
   ↓
(3) Deeper EDA ← now data is more reliable
   ↓
(4) Feature engineering
   ↓
(5) Modeling
   ↓
(6) Error analysis → more EDA again

🧭 A. BEFORE cleaning (Exploratory understanding)

You try to answer:

What does each column mean?
What is missing?
Are there weird values?
What is the target distribution?
Are there duplicates / broken rows?

👉 This is “diagnostic EDA”

🧹 B. DURING cleaning (decision-making EDA)

You use EDA to decide:

Should I remove Open = 0?
Should I drop outliers?
How should I impute missing values?
Is this feature skewed?

👉 This is “action EDA”

🚀 C. AFTER cleaning (modeling EDA)

Now you analyze:

Correlation with target
Seasonality patterns
Store-level behavior
Promo effects

👉 This is “feature engineering EDA”
